## `03:Baum`
Füge einen Baumstamm hinzu:
```txt
         *          
        ***         
       *****        
      *******       
     *********      
    ***********     
   *************    
  ***************   
 *****************  
******************* 
         **         
         **         
         **         
```

````python
# Prüfen (optional): führe diese Zelle aus, um deine Lösung zu testen.
import io, contextlib
_check_buf = io.StringIO()
with contextlib.redirect_stdout(_check_buf):
    exec(_ih[-2])
_lines = [l for l in _check_buf.getvalue().split(chr(10)) if l.strip() != '']
_counts = [l.count('*') for l in _lines]
_expected = list(range(1, 20, 2)) + [2, 2, 2]
assert _counts == _expected, f"Erwarte die Pyramide (1,3,...,19 Sterne) gefolgt von 3 Zeilen mit je 2 Sternen (Stamm), aber bekommen: {_counts}"
print("Richtig!")
````
